```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# =====================================================================
# ÉTAPE 1 : COLLECTE ET CRÉATION DE LA BASE DE DONNÉES (VERSION ÉTENDUE)
# =====================================================================
donnees_brutes = {
    "contenu": [
        "Gagnez 1000 euros facilement en cliquant ici !",
        "Bonjour, la réunion est déplacée à demain 14h.",
        "Félicitations ! Votre compte a été sélectionné pour un cadeau.",
        "Est-ce que tu es disponible pour un café ce midi ?",
        "URGENT: Modifiez votre mot de passe immédiatement.",
        "Merci pour votre achat, voici votre facture en pièce jointe.",
        "Offre exclusive : gagnez des cadeaux gratuits dès maintenant !",
        "Salut, tu as pensé à appeler le plombier pour la fuite ?",
        "Doublez vos revenus en travaillant depuis chez vous.",
        "Votre colis a été expédié, suivez sa livraison ici.",
        "⚠️ ALERTE SÉCURITÉ: Tentative de connexion suspecte sur votre compte.",
        "Peux-tu m'envoyer le compte-rendu de la réunion s'il te plaît ?",
        "Promotion incroyable sur les crypto-monnaies achetez vite !",
        "Je serai un peu en retard pour le dîner ce soir, désolé.",
        "Gagnez un iPhone 15 Pro en répondant à ce sondage rapide.",
        "Le point sur le projet de fin d'année est disponible sur le drive.",
    ],
    "type_email": [
        "spam",
        "normal",
        "spam",
        "normal",
        "spam",
        "normal",
        "spam",
        "normal",
        "spam",
        "normal",
        "spam",
        "normal",
        "spam",
        "normal",
        "spam",
        "normal",
    ],
}
tableau_donnees = pd.DataFrame(donnees_brutes)


# =====================================================================
# ÉTAPE 2 : AFFICHAGE DES DONNÉES BRUTES & NETTOYAGE
# =====================================================================
print("=== 1. AFFICHAGE DES DONNÉES BRUTES ===")
print(tableau_donnees.head())

# VRAI NETTOYAGE
tableau_donnees = tableau_donnees.dropna(subset=["contenu"])  # Supprime le vide
tableau_donnees = tableau_donnees.drop_duplicates(
    subset=["contenu"]
)  # Supprime les doublons
tableau_donnees["contenu"] = tableau_donnees[
    "contenu"
].str.lower()  # Tout en minuscules

print(
    f"\nDonnées nettoyées ! Lignes restantes : {len(tableau_donnees)}"
)
print("\n" + "=" * 50 + "\n")


# =====================================================================
# ÉTAPE 3 : ANALYSE ET STATISTIQUES (VISUALISATION)
# =====================================================================
comptage_categories = tableau_donnees["type_email"].value_counts()
print("=== 2. COMPTAGE DES CATÉGORIES ===")
print(comptage_categories)

# Dessin du graphique
plt.figure(figsize=(5, 3))
plt.bar(
    comptage_categories.index,
    comptage_categories.values,
    color=["#4CAF50", "#FF5722"],
)
plt.title("Répartition des E-mails")
plt.ylabel("Nombre d'e-mails")
plt.show()


# =====================================================================
# ÉTAPE 4 : ENCODAGE ET TRADUCTION NUMÉRIQUE (PRÉPARATION)
# =====================================================================
# A. Traduction de l'étiquette (normal -> 0, spam -> 1)
convertisseur_label = LabelEncoder()
tableau_donnees["type_encode"] = convertisseur_label.fit_transform(
    tableau_donnees["type_email"]
)

# B. Traduction du texte en matrice de nombres (TF-IDF)
traducteur_texte = TfidfVectorizer()
X = traducteur_texte.fit_transform(tableau_donnees["contenu"])
y = tableau_donnees["type_encode"]


# =====================================================================
# ÉTAPE 5 : SÉPARATION DES DONNÉES (ENTRAÎNEMENT vs TEST)
# =====================================================================
X_entrainement, X_test, y_entrainement, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)


# =====================================================================
# ÉTAPE 6 : ENTRAÎNEMENT DE L'ALGORITHME (MESSAGERIE IA)
# =====================================================================
modele_ia = LogisticRegression()
modele_ia.fit(X_entrainement, y_entrainement)
print("\n=== 3. ENTRAÎNEMENT DE L'IA RÉUSSI ===\n")


# =====================================================================
# ÉTAPE 7 : TESTS ET ÉVALUATION DES PERFORMANCES
# =====================================================================
predictions = modele_ia.predict(X_test)
taux_reussite = accuracy_score(y_test, predictions)

print(f"Précision globale du modèle : {taux_reussite * 100:.1f} %")
print("\nRapport d'évaluation détaillé :")
print(
    classification_report(
        y_test, predictions, target_names=["Normal", "Spam"]
    )
)
print("\n" + "=" * 50 + "\n")


# =====================================================================
# APPLICATION CONCRÈTE : TEST EN DIRECT
# =====================================================================
print("=== 4. TEST EN DIRECT AVEC UN NOUVEL E-MAIL ===")

# Vous pouvez modifier cette phrase pour tester l'IA :
nouvel_email = ["URGENT ! Cliquez ici pour recevoir votre virement gratuit"]

# Encodage obligatoire du nouvel e-mail avec le même traducteur
nouvel_email_encode = traducteur_texte.transform(nouvel_email)

# Verdict de l'IA
verdict = modele_ia.predict(nouvel_email_encode)
probabilites = modele_ia.predict_proba(nouvel_email_encode)
pourcentage_certitude = probabilites[0][verdict[0]] * 100

if verdict[0] == 1:
    print(
        f"Résultat : [SPAM DETECTÉ] (Certitude : {pourcentage_certitude:.1f}%)"
    )
else:
    print(
        f"Résultat : [MESSAGE SÛR] (Certitude : {pourcentage_certitude:.1f}%)"
    )
```

```python
import torch
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments

# 1. Chargement ultra-rapide des outils
dataset = load_dataset("stanfordnlp/imdb")
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

# 2. Tokenisation des textes
def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True)

tokenized_datasets = dataset.map(tokenize_function, batched=True)

# 3. Chargement du modèle BERT (DistilBERT)
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=2)

# 4. Paramètres d'entraînement configurés pour votre Tesla T4
training_args = TrainingArguments(
    output_dir="./results",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    num_train_epochs=1,
    weight_decay=0.01,
    logging_steps=5,     # Affiche un rapport toutes les 5 étapes pour voir l'évolution
    fp16=True            # Option cruciale qui exploite la puissance brute du T4
)

# 5. Création du Trainer avec un échantillon léger pour valider le fonctionnement
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"].shuffle(seed=42).select(range(400)),
    eval_dataset=tokenized_datasets["test"].shuffle(seed=42).select(range(100)),
)

# 6. Lancement de l'apprentissage !
trainer.train()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms

# 1. Vérification matérielle immédiate
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"--- VÉRIFICATION : Le code tourne sur [{device}] ---")

# 2. Chargement simplifié des données (num_workers=0 pour éviter les blocages)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=0)

# 3. Réseau de neurones Convolutif (CNN)
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(2, 2)
        self.fc = nn.Linear(16 * 16 * 16, 10)

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = x.view(x.size(0), -1)
        return self.fc(x)

# 4. Initialisation et entraînement
model = SimpleCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("\nLancement de la boucle d'apprentissage...")
for epoch in range(1):
    for i, (inputs, labels) in enumerate(trainloader):
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        if i % 100 == 99:
            avancement = (i / len(trainloader)) * 100
            print(f"Progression : {avancement:.0f}% | Perte : {loss.item():.4f}")

print("Félicitations ! Entraînement terminé en quelques secondes.")


```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.models as models
import ipywidgets as widgets
from IPython.display import display, clear_output

# 1. Configuration de l'environnement GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Configuration : Modèle chargé sur [{device}]")

# 2. Génération de fausses images en mémoire pour simuler les déchets
# 1000 images, 3 canaux (RVB), taille 224x224 (taille standard pour ResNet)
print("Génération des images de déchets virtuels en cours...")
X_dechets = torch.randn(1000, 3, 224, 224)
# 4 classes : 0=Plastique, 1=Verre, 2=Carton, 3=Métal
y_dechets = torch.randint(0, 4, (1000,))

dataset = torch.utils.data.TensorDataset(X_dechets, y_dechets)
trainloader = torch.utils.data.DataLoader(dataset, batch_size=32, shuffle=True)

# 3. TRANSFER LEARNING : Initialisation de ResNet50
print("Chargement du modèle pré-entraîné ResNet50...")
# weights=models.ResNet50_Weights.DEFAULT charge les poids entraînés sur ImageNet
resnet = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)

# On gèle les couches de base pour conserver leur intelligence visuelle
for param in resnet.parameters():
    param.requires_grad = False

# On remplace la dernière couche linéaire (fc) pour l'adapter à nos 4 classes
num_features = resnet.fc.in_features
resnet.fc = nn.Linear(num_features, 4)

model_dechets = resnet.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_dechets.fc.parameters(), lr=0.005) # On n'entraîne QUE la dernière couche

# 4. Interface Graphique Widgets
titre = widgets.HTML("<h2>♻️ IA de Tri Automatique des Déchets (Transfer Learning)</h2>")
bouton_train = widgets.Button(description="🏋️ Entraîner l'IA (1 époque)", button_style='warning', layout=widgets.Layout(width='250px'))
bouton_test = widgets.Button(description="📸 Simuler un scan de déchet", button_style='info', layout=widgets.Layout(width='250px'))
zone_resultat = widgets.Output()

def entrainer_ia(b):
    with zone_resultat:
        clear_output()
        print("⚡ Entraînement flash de la couche supérieure sur GPU...")
        model_dechets.train()
        for inputs, labels in trainloader:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model_dechets(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
        print("✅ Entraînement terminé ! La couche de tri est ajustée.")

def tester_ia(b):
    with zone_resultat:
        clear_output()
        model_dechets.eval()
        # Simulation du scan d'une nouvelle image aléatoire
        fausse_image = torch.randn(1, 3, 224, 224).to(device)
        with torch.no_grad():
            output_pred = model_dechets(fausse_image)
            _, prediction = torch.max(output_pred, 1)

        classes = ["Plastique 🛢️", "Verre 🍾", "Carton 📦", "Métal 🥫"]
        resultat_final = classes[prediction.item()]

        print("="*50)
        print(f" 📦 OBJET SCANNÉ SUR LE TAPIS ROULANT")
        print(f" DIRECTION LE BAC : {resultat_final}")
        print("="*50)

bouton_train.on_click(entrainer_ia)
bouton_test.on_click(tester_ia)

clear_output()
display(titre, widgets.HBox([bouton_train, bouton_test]), zone_resultat)
```

```python
import torch
import ipywidgets as widgets
from IPython.display import display, clear_output
from PIL import Image, ImageDraw
from transformers import BlipProcessor, BlipForConditionalGeneration

# 1. Configuration et chargement de l'I.A. multimodale BLIP
print("Chargement de l'IA multimodale (BLIP Image-to-Text)...")
device = "cuda" if torch.cuda.is_available() else "cpu"

processeur = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
modele_hybride = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base").to(device)
print(f"Modèle hybride prêt sur [{device}] !")

# 2. Interface Graphique Widgets
titre = widgets.HTML("<h2>📸 Générateur de Légendes Automatiques (Multimodal IA)</h2>")
description = widgets.Label("Cliquez sur le bouton pour faire analyser l'image générée en mémoire :")
bouton_generer = widgets.Button(description="🧠 Analyser l'image & légender", button_style='success', layout=widgets.Layout(width='250px'))
zone_resultat = widgets.Output()

def analyser_image_multimodale(b):
    with zone_resultat:
        clear_output()
        print("🎨 Création d'une image d'un carré rouge sur fond bleu en mémoire...")

        # On crée une vraie image PIL de 224x224 pixels (fond bleu)
        image = Image.new("RGB", (224, 224), "blue")
        # On y dessine un grand carré rouge au centre
        draw = ImageDraw.Draw(image)
        draw.rectangle([60, 60, 160, 160], fill="red")

        print("🔮 L'IA analyse les formes et génère la légende textuelle...")
        # Transformation des pixels pour le modèle
        inputs = processeur(images=image, return_tensors="pt").to(device)

        # Génération automatique de la description textuelle
        with torch.no_grad():
            outputs = modele_hybride.generate(**inputs)
            phrase_generee = processeur.decode(outputs, skip_special_tokens=True)

        print("\n" + "="*50)
        print(f" 🖼️ IMAGE SCANNÉE : [Carré rouge dessiné sur fond bleu]")
        print(f" 📝 LÉGENDE GÉNÉRÉE : {phrase_generee}")
        print("="*50)

bouton_generer.on_click(analyser_image_multimodale)

clear_output()
display(titre, description, bouton_generer, zone_resultat)

```

```python
# =====================================================================
# PROJET IA COMPLET : CLASSIFICATION AUTOMATIQUE DE TEXTES
# =====================================================================

# 1. IMPORTATION DES BIBLIOTHÈQUES
print("1. Importation des outils...")
import string
import pandas as pd
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# 2. COLLECTE ET LECTURE DES DONNÉES
print("2. Collecte des données en cours...")
# On télécharge deux catégories distinctes : l'informatique (comp.graphics) et la religion (soc.religion.christian)
categories = ['comp.graphics', 'soc.religion.christian']
donnees_brutes = fetch_20newsgroups(subset='all', categories=categories, remove=('headers', 'footers', 'quotes'))

# Création du tableau de données (DataFrame)
df = pd.DataFrame({
    'Texte': donnees_brutes.data,
    'Categorie_Id': donnees_brutes.target
})
# Association de l'ID à un nom clair (0 = Informatique, 1 = Religion)
df['Nom_Categorie'] = df['Categorie_Id'].apply(lambda x: donnees_brutes.target_names[x])

print(f"   -> {len(df)} textes récupérés avec succès.")

# 3. NETTOYAGE ET PRÉPARATION DES DONNÉES (Gestion de la casse et ponctuation)
print("3. Nettoyage du texte (casse et ponctuation)...")
def nettoyer_texte(texte):
    if not isinstance(texte, str): return ""
    texte = texte.lower()  # Passage en minuscules
    table_ponctuation = str.maketrans('', '', string.punctuation)
    texte = texte.translate(table_ponctuation)  # Retrait ponctuation
    return " ".join(texte.split())  # Suppression des espaces superflus

df['Texte_Nettoye'] = df['Texte'].apply(nettoyer_texte)

# Séparation des données pour l'entraînement (80%) et les tests (20%)
X_train, X_test, y_train, y_test = train_test_split(
    df['Texte_Nettoye'], df['Categorie_Id'], test_size=0.20, random_state=42
)

# 4. ENCODAGE DES DONNÉES (Transformation du texte en chiffres)
print("4. Encodage numérique des textes...")
vectorizer = TfidfVectorizer(max_features=2000)
X_train_encoded = vectorizer.fit_transform(X_train)
X_test_encoded = vectorizer.transform(X_test)

# 5. ENTRAÎNEMENT DU MODÈLE d'IA
print("5. Entraînement de l'algorithme d'IA (Naive Bayes)...")
modele_ia = MultinomialNB()
modele_ia.fit(X_train_encoded, y_train)

# 6. TESTS ET ÉVALUATION
print("6. Évaluation des performances du modèle...")
predictions = modele_ia.predict(X_test_encoded)
precision = accuracy_score(y_test, predictions)

print("\n================== RÉSULTATS DU PROJET ==================")
print(f"Précision globale du modèle d'IA : {precision:.2%}")
print("\nDétail des performances par catégorie :")
print(classification_report(y_test, predictions, target_names=donnees_brutes.target_names))
print("=========================================================\n")

# 7. DÉPLOIEMENT / PREDICTION SUR DE NOUVELLES DONNÉES (Simulation d'usage réel)
print("7. Simulation de déploiement (Test en conditions réelles) :")

phrases_inconnues = [
    "I need to render a 3D pixel design using a new graphics card software.",
    "The priest prayed for peace during the Sunday morning church service."
]

# Nettoyage et encodage des nouvelles phrases
phrases_nettoyees = [nettoyer_texte(p) for p in phrases_inconnues]
phrases_encodees = vectorizer.transform(phrases_nettoyees)

# Prédiction de l'IA
resultats_predictions = modele_ia.predict(phrases_encodees)

for phrase, pred_id in zip(phrases_inconnues, resultats_predictions):
    categorie_predite = donnees_brutes.target_names[pred_id]
    print(f"\n-> Phrase analysée : '{phrase}'")
    print(f"   IA prédit la catégorie : [ {categorie_predite} ]")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import xlrd
import os

# 1. Détection automatique du fichier Excel d'origine
fichiers_excel = [f for f in os.listdir('.') if f.endswith(('.xlsx', '.xls')) and not f.startswith('manifeste_')]

if not fichiers_excel:
    print("❌ Aucun fichier trouvé. Assurez-vous que votre fichier est bien présent dans le panneau de gauche 📂.")
else:
    nom_du_fichier_real = fichiers_excel[0]
    print(f"⚙️ Analyse du fichier : '{nom_du_fichier_real}'")

    # 2. Lecture sécurisée du manifeste
    try:
        workbook = xlrd.open_workbook(nom_du_fichier_real, ignore_workbook_corruption=True)
        df = pd.read_excel(workbook)
    except Exception as e:
        df = pd.read_excel(nom_du_fichier_real)

    # 3. Règle de validation douanière et logistique
    def valider_marchandise(row):
        if 'Tonnage (Kg)' in row and pd.notna(row['Tonnage (Kg)']):
            try:
                poids = float(row['Tonnage (Kg)'])
                if poids <= 0: return "Alerte: Tonnage nul ou négatif"
            except:
                return "Erreur: Format du Tonnage invalide"
        if 'N° BL' in row and (pd.isna(row['N° BL']) or str(row['N° BL']).strip() == ""):
            return "Erreur: N° BL manquant"
        return "OK"

    # 4. Nettoyage et conversion des données
    for col in df.select_dtypes(include=['object']).columns:
        df[col] = df[col].astype(str).str.strip()

    if 'Tonnage (Kg)' in df.columns:
        df['Tonnage (Kg)'] = pd.to_numeric(df['Tonnage (Kg)'], errors='coerce')

    df['Statut_Pipeline'] = df.apply(valider_marchandise, axis=1)

    # 5. Séparation en sous-ensembles (Filtres)
    df_ok = df[df['Statut_Pipeline'] == "OK"]
    df_erreurs = df[df['Statut_Pipeline'] != "OK"]

    # 6. Calcul des statistiques par nature de cargaison
    if 'Nature M/se' in df.columns and 'Tonnage (Kg)' in df.columns:
        stats_marchandise = df.groupby('Nature M/se')['Tonnage (Kg)'].agg(['count', 'sum']).reset_index()
        stats_marchandise.columns = ['Nature de la Marchandise', 'Nombre de BL', 'Tonnage Total (Kg)']
        stats_marchandise = stats_marchandise.sort_values(by='Tonnage Total (Kg)', ascending=False)
    else:
        stats_marchandise = pd.DataFrame({'Message': ['Colonnes requises absentes du fichier']})

    # 7. Écriture du fichier Excel final multi-onglets
    nom_fichier_sortie = 'manifeste_port_organise.xlsx'
    with pd.ExcelWriter(nom_fichier_sortie, engine='openpyxl') as writer:
        df_ok.to_excel(writer, sheet_name='VALIDE - OK', index=False)
        df_erreurs.to_excel(writer, sheet_name='A CORRIGER', index=False)
        stats_marchandise.to_excel(writer, sheet_name='STATISTIQUES', index=False)

    print(f"🎉 Pipeline terminé avec succès !")
    print(f"📁 Le fichier '{nom_fichier_sortie}' a été créé dans le dossier de gauche (📂).")

    # 8. Génération visuelle du graphique de fret
    if 'Nature de la Marchandise' in stats_marchandise.columns and len(stats_marchandise) > 0:
        print("\n📊 Génération du tableau de bord visuel :")
        plt.figure(figsize=(10, 5))
        sns.barplot(
            x='Tonnage Total (Kg)',
            y='Nature de la Marchandise',
            data=stats_marchandise,
            palette='Blues_r',
            hue='Nature de la Marchandise',
            legend=False
        )
        plt.title('Répartition du Tonnage par Nature de Marchandise', fontsize=14, weight='bold', pad=15)
        plt.xlabel('Tonnage Total (Kg)', fontsize=11)
        plt.ylabel('Nature de la Marchandise', fontsize=11)
        plt.grid(axis='x', linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

        # Affichage du tableau récapitulatif
        display(stats_marchandise)
```

```python
import pandas as pd
import xlrd
import os
import shutil

# 1. Détection automatique du nom du fichier (Correction de la liste)
fichiers_excel = [f for f in os.listdir('.') if f.endswith(('.xlsx', '.xls')) and not f.startswith('rapport_') and not f.startswith('manifeste_')]

if not fichiers_excel:
    print("❌ Aucun fichier trouvé. Assurez-vous que votre fichier est bien dans le panneau de gauche 📂.")
else:
    # On prend le premier fichier trouvé et on extrait la chaîne de caractères (pas la liste)
    nom_du_fichier_real = fichiers_excel[0]
    print(f"⚙️ Analyse statistique avancée du fichier : '{nom_du_fichier_real}'")

    # 2. Lecture sécurisée du manifeste avec xlrd
    try:
        workbook = xlrd.open_workbook(nom_du_fichier_real, ignore_workbook_corruption=True)
        df = pd.read_excel(workbook)
    except Exception as e:
        df = pd.read_excel(nom_du_fichier_real)

    # 3. Nettoyage initial des données
    for col in df.select_dtypes(include=['object']).columns:
        df[col] = df[col].astype(str).str.strip()
    if 'Tonnage (Kg)' in df.columns:
        df['Tonnage (Kg)'] = pd.to_numeric(df['Tonnage (Kg)'], errors='coerce').fillna(0)
    if 'Qté' in df.columns:
        df['Qté'] = pd.to_numeric(df['Qté'], errors='coerce').fillna(0)

    # ==========================================================
    # INDIFICATEUR 1 : Conteneurs par Pays en Transit (Hors Togo)
    # ==========================================================
    if 'Type Emballage' in df.columns and 'Pays Dest.' in df.columns:
        # On filtre sur l'Afrique Centrale/Périphérique en transit (ex: Gabon présent dans vos données)
        df_transit = df[(df['Pays Dest.'].str.upper() != 'TOGO') & (df['Pays Dest.'].str.upper() != 'TG')]
        if len(df_transit) > 0:
            stats_transit = df_transit.groupby('Pays Dest.')['Qté'].sum().reset_index()
            stats_transit.columns = ['Pays de Destination', 'Nombre total de Colis/Conteneurs']
            stats_transit = stats_transit.sort_values(by='Nombre total de Colis/Conteneurs', ascending=False)
        else:
            stats_transit = pd.DataFrame({'Message': ['Aucune marchandise internationale en transit direct détectée']})
    else:
        stats_transit = pd.DataFrame({'Message': ['Colonnes logistiques manquantes pour le transit']})

    # ==========================================================
    # INDICATEUR 2 : Classement par Nature de Marchandise Import
    # ==========================================================
    if 'Nature M/se' in df.columns and 'Tonnage (Kg)' in df.columns:
        stats_import = df.groupby('Nature M/se')['Tonnage (Kg)'].sum().reset_index()
        stats_import.columns = ['Nature de la Marchandise', 'Tonnage Cumulé (Kg)']
        stats_import = stats_import.sort_values(by='Tonnage Cumulé (Kg)', ascending=False)
    else:
        stats_import = pd.DataFrame({'Message': ['Colonnes Nature M/se ou Tonnage manquantes']})

    # ==========================================================
    # INDICATEUR 3 : Statistiques des Navires par Pavillon
    # ==========================================================
    if 'Navire' in df.columns and 'Pays Prov.' in df.columns:
        # À défaut d'une colonne pavillon dédiée, le Pays de Provenance de l'armement ou de la ligne sert d'indicateur
        stats_navires = df.groupby('Pays Prov.')['Navire'].nunique().reset_index()
        stats_navires.columns = ['Pavillon / Origine de la Ligne', 'Nombre de Navires Uniques']
        stats_navires = stats_navires.sort_values(by='Nombre de Navires Uniques', ascending=False)
    else:
        stats_navires = pd.DataFrame({'Message': ['Colonnes Navire ou Pays Prov. manquantes']})

    # ==========================================================
    # 4. ENREGISTREMENT ET EXPORTATION MULTI-ONGLETS
    # ==========================================================
    nom_fichier_complet = 'rapport_statistiques_portuaires.xlsx'
    with pd.ExcelWriter(nom_fichier_complet, engine='openpyxl') as writer:
        df.to_excel(writer, sheet_name='DONNEES_GLOBALES', index=False)
        stats_transit.to_excel(writer, sheet_name='TRANSIT_PAR_PAYS', index=False)
        stats_import.to_excel(writer, sheet_name='CLASSEMENT_IMPORT', index=False)
        stats_navires.to_excel(writer, sheet_name='STAT_PAVILLONS', index=False)

    print(f"🎉 Pipeline exécuté avec succès ! Fichier généré localement : '{nom_fichier_complet}'")

    # 5. COPIE DE SÉCURITÉ AUTOMATIQUE VERS GOOGLE DRIVE
    dossier_destination = '/content/drive/MyDrive/Pipeline_Manifestes'
    if os.path.exists(dossier_destination):
        chemin_final = os.path.join(dossier_destination, 'Rapport_Analytique_Mars_2026.xlsx')
        shutil.copy(nom_fichier_complet, chemin_final)
        print(f"🚀 Version finale mise à jour sur Google Drive : 'Pipeline_Manifestes > Rapport_Analytique_Mars_2026.xlsx'")

    # Affichage des aperçus à l'écran
    print("\n🌍 Aperçu - Transit par Pays :")
    display(stats_transit.head(3))
    print("\n👑 Aperçu - Top Imports :")
    display(stats_import.head(3))



```

```python
import pandas as pd
import xlrd
import os
import shutil

# 1. Détection automatique du nom du fichier
fichiers_excel = [f for f in os.listdir('.') if f.endswith(('.xlsx', '.xls')) and not f.startswith('rapport_') and not f.startswith('manifeste_')]

if fichiers_excel:
    nom_du_fichier_real = fichiers_excel[0]

    # Lecture du fichier
    try:
        workbook = xlrd.open_workbook(nom_du_fichier_real, ignore_workbook_corruption=True)
        df = pd.read_excel(workbook)
    except:
        df = pd.read_excel(nom_du_fichier_real)

    # Nettoyage
    for col in df.select_dtypes(include=['object']).columns:
        df[col] = df[col].astype(str).str.strip()
    df['Tonnage (Kg)'] = pd.to_numeric(df['Tonnage (Kg)'], errors='coerce').fillna(0)

    # ==========================================================
    # 1. STATISTIQUES DES PAVILLONS + PARTS DE MARCHÉ (%)
    # ==========================================================
    if 'Navire' in df.columns and 'Pays Prov.' in df.columns:
        stats_navires = df.groupby('Pays Prov.')['Navire'].nunique().reset_index()
        stats_navires.columns = ['Pavillon / Origine', 'Nombre de Navires Uniques']

        # Calcul du pourcentage de part de marché (Part des escales)
        total_navires = stats_navires['Nombre de Navires Uniques'].sum()
        stats_navires['Part de Marché (%)'] = ((stats_navires['Nombre de Navires Uniques'] / total_navires) * 100).round(2)
        stats_navires = stats_navires.sort_values(by='Nombre de Navires Uniques', ascending=False)
    else:
        stats_navires = pd.DataFrame({'Message': ['Colonnes manquantes']})

    # ==========================================================
    # 2. ESTIMATION DES TAXES DE MANUTENTION (EXEMPLE DE BAREME)
    # ==========================================================
    # Simulation d'un barème de manutention (ex: 5 FCFA par Kg pour le vrac/conteneur standard)
    # Vous pourrez ajuster ce tarif (0.005 ici soit 5 FCFA par Kg) selon vos vraies grilles tarifaires
    TARIF_MANUTENTION_PAR_KG = 0.005

    df['Taxe_Manutention_Estimee_FCFA'] = (df['Tonnage (Kg)'] * TARIF_MANUTENTION_PAR_KG).round(0)

    # Calcul du résumé financier par Manutentionnaire (LCT vs Port Autonome de Lomé)
    if 'Manutentionnaire' in df.columns:
        stats_taxes = df.groupby('Manutentionnaire')['Taxe_Manutention_Estimee_FCFA'].sum().reset_index()
        stats_taxes.columns = ['Opérateur / Manutentionnaire', 'Total Taxes Estimées (FCFA)']
        # Formatage lisible des montants
        stats_taxes['Total Taxes Estimées (FCFA)'] = stats_taxes['Total Taxes Estimées (FCFA)'].apply(lambda x: "{:,.0f}".format(x).replace(',', ' '))
    else:
        stats_taxes = pd.DataFrame({'Message': ['Colonne Manutentionnaire manquante']})

    # ==========================================================
    # 3. SAUVEGARDE ET EXPORTATION MULTI-ONGLETS ENRICHIE
    # ==========================================================
    nom_fichier_complet = 'rapport_statistiques_portuaires.xlsx'
    with pd.ExcelWriter(nom_fichier_complet, engine='openpyxl') as writer:
        df.to_excel(writer, sheet_name='DONNEES_GLOBALES_TAXES', index=False)
        stats_navires.to_excel(writer, sheet_name='PARTS_MARCHE_PAVILLONS', index=False)
        stats_taxes.to_excel(writer, sheet_name='ESTIMATION_TAXES', index=False)

    # Copie vers Google Drive
    dossier_destination = '/content/drive/MyDrive/Pipeline_Manifestes'
    if os.path.exists(dossier_destination):
        chemin_final = os.path.join(dossier_destination, 'Rapport_Analytique_Mars_2026.xlsx')
        shutil.copy(nom_fichier_complet, chemin_final)
        print("✅ Rapport financier et parts de marché mis à jour sur votre Google Drive !")

    # Affichage des aperçus
    print("\n🚢 Parts de marché par Pavillon / Origine des Navires :")
    display(stats_navires.head(5))
    print("\n💰 Estimation des recettes de manutention par opérateur (Base: 5 FCFA / Kg) :")
    display(stats_taxes)
```

```python
import pandas as pd
import yfinance as yf
import datetime

# ==========================================
# 1. EXTRACT (Extraction des données)
# ==========================================
def extract_financial_data(ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
    """Télécharge les données depuis yfinance et simplifie la structure des colonnes."""
    print(f"--- [EXTRACT] Téléchargement des données pour {ticker} ---")

    # Correction du FutureWarning avec auto_adjust=True
    data = yf.download(ticker, start=start_date, end=end_date, auto_adjust=True)

    # Si yfinance renvoie un MultiIndex (ex: colonnes imbriquées avec le Ticker)
    if isinstance(data.columns, pd.MultiIndex):
        # On ne garde que le premier niveau (Close, Open, etc.)
        data.columns = data.columns.get_level_values(0)

    return data

# ==========================================
# 2. TRANSFORM (Nettoyage et Calculs)
# ==========================================
def transform_data(df: pd.DataFrame) -> pd.DataFrame:
    """Nettoie le DataFrame et calcule des indicateurs financiers."""
    print("--- [TRANSFORM] Nettoyage et calcul des indicateurs ---")

    # Copie pour éviter les avertissements de modification
    df = df.copy()

    # 1. Gestion des valeurs manquantes (au cas où)
    df = df.ffill().bfill()

    # 2. Calcul du rendement quotidien (Daily Return %)
    # Formule : (Prix_J / Prix_J-1) - 1
    df['Daily_Return_%'] = df['Close'].pct_change() * 100

    # 3. Calcul de la Moyenne Mobile Simple sur 20 jours (SMA 20)
    df['SMA_20'] = df['Close'].rolling(window=20).mean()

    # 4. Calcul de la Volatilité Mobile sur 20 jours (Écart-type des rendements)
    df['Volatility_20d'] = df['Daily_Return_%'].rolling(window=20).std()

    # 5. Supprimer les lignes avec des NaN (générés par la fenêtre de 20 jours)
    df = df.dropna()

    return df

# ==========================================
# 3. LOAD (Sauvegarde / Chargement)
# ==========================================
def load_data(df: pd.DataFrame, output_filename: str):
    """Sauvegarde les données transformées dans un fichier CSV ou base de données."""
    print(f"--- [LOAD] Sauvegarde des données dans {output_filename} ---")
    df.to_csv(output_filename)
    print("Pipeline exécuté avec succès ! 🎉")

# ==========================================
# EXÉCUTION DU PIPELINE (Orchestration)
# ==========================================
if __name__ == "__main__":
    # Configuration des paramètres
    TICKER = "AAPL"
    START = "2025-01-01"
    END = "2025-12-31"
    OUTPUT = "apple_transformed_data.csv"

    # Exécution pas à pas du pipeline
    raw_data = extract_financial_data(TICKER, START, END)
    processed_data = transform_data(raw_data)
    load_data(processed_data, OUTPUT)

    # Visualisation d'un échantillon du résultat final
    print("\n--- Aperçu des données prêtes pour l'analyse ---")
    print(processed_data[['Close', 'Daily_Return_%', 'SMA_20', 'Volatility_20d']].tail())
```
